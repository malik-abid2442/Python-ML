# Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)

X = np.random.rand(50,5, 1) * 100  

Y = 3.5 * X + np.random.randn(50,5, 1) * 20
```

```python
plt.scatter(X,Y)
```

```python
Y = Y.reshape(-1)
model = LinearRegression()
model.fit(X,Y)
```

```python
print("X shape:", X.shape)
print("Y shape:", Y.shape)

print("X ndim:", X.ndim)
print("Y ndim:", Y.ndim)
```

```python
y_pred = model.predict(X)
```

```python
plt.figure(figsize=(8,6)) 
plt.scatter(X, Y, color='blue', label='Data Points') 
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line') 
plt.title('Linear Regression on Random Dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
size = [800, 900, 1000, 1100, 1200, 1400, 1500, 1600, 1800, 2000]

price = [145000, 160000, 178000, 190000, 215000, 245000, 260000, 278000, 320000, 350000]

X = np.array(size)
Y = np.array(price)
```

```python
X,Y
```

```python
model = LinearRegression()
X = X.reshape(-1, 1)
Y = Y.reshape(-1)
model.fit(X,Y)
```

```python
y_pred= model.predict(X)
```

```python
model.r2_score()
```

```python
model.score(X, Y)
```

```python
from sklearn.metrics import r2_score

y_pred = model.predict(X)

r2 = r2_score(Y, y_pred)

print(r2)
```
