# Iris Flower Classification

```python
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd 
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=1)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
linear = SVC(kernel='linear',C=1)
linear.fit(X_train,y_train)
acc = accuracy_score(y_test, linear.predict(X_test)) * 100
print(f"Linear model accuracy: {acc:.2f}%")
```

```python
rbf = SVC(kernel='rbf', C=2)
rbf.fit(X_train,y_train)
acc = accuracy_score(y_test, rbf.predict(X_test)) * 100
print(f"rbf model accuracy: {acc:.2f}%")
```

# Breast Cancer Classification

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
X = cancer.data
y = cancer.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=1)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
C = [0.1, 1, 10, 100]
for c in C:
    linear = SVC(kernel='linear',C=c)
    linear.fit(X_train,y_train)
    acc = accuracy_score(y_test, linear.predict(X_test)) * 100
    print(f"Linear model accuracy: {acc:.2f}%")
    print('Value of c : ',c)
```

```python
C = [0.1, 1, 10, 100]
gamma = [0.001, 0.01, 0.1, 1]
for c in C:
    for g in gamma:
        rbf = SVC(kernel='rbf',C=c,gamma=g)
        rbf.fit(X_train,y_train)
        acc = accuracy_score(y_test, rbf.predict(X_test)) * 100
        print(f"rbf model accuracy: {acc:.2f}%")
        print('Value of c : ',c)
        print('Value of Gamma : ',g)
```
