# Exploratiory Data Analysis

```python
import numpy as np
import pandas as pd 
import seaborn as sns
```

```python
#load Data Through Seaborn
df = sns.load_dataset("titanic")
```

# 1. Categorical Data

```python
sns.countplot(df['alive'])
```

```python
df['sex'].value_counts().plot(kind='pie',autopct='%.2f')
```

# 2. Numerical Data

```python
# Histogram
sns.histplot(df['age'],kde=True,bins=80)
```

```python
# Boxplot
sns.boxplot(df['age'])
```

# Bivariate Analysis

```python
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
flights = sns.load_dataset('flights')
```

```python
# Scatterplot => Num - Num
sns.scatterplot(x='total_bill', y='tip', data=tips,hue='sex',style='smoker')
```

```python
# Barplot => num - cat
sns.barplot(x='pclass',y='fare',data=df,hue='survived')
```

```python
# boxplot => num - cat 
sns.boxplot(x='sex',y='age',data=df,hue='survived')
```

```python
# Displot => num - cat
sns.displot(x='age',data=df,hue='survived',kde=True)
```

```python
# Heatmap => cat - cat
sns.heatmap(pd.crosstab(df['pclass'],df['survived']))
```

```python
df.groupby('pclass')['survived'].mean()
```

```python
sns.pairplot(iris,hue='species')
```
