```python
class student:
    def __init__(self,name,marks,rollno):
        self.name=name
        self.marks=marks
        self.rollno=rollno
    def display_student(self):
        print("Name:",self.name)
        print("Marks:",self.marks)
        print("Roll No:",self.rollno)
    def calculate_grade(self):
        if self.marks>=90:
            return "A"
        elif self.marks>=80:
            return "B"
        elif self.marks>=70:
            return "C"
        elif self.marks>=60:
            return "D"
        else:
            return "F"

abid = student("Abid", 85, 101)
abid.display_student()
abids_grade = abid.calculate_grade()
print("Grade:", abids_grade)
        
```

```python
class car:
    def __init__(self,brand,model,year):
        self.brand=brand
        self.model=model
        self.year=year

    def start_engine(self):
        print(f"The engine of {self.year} {self.brand} {self.model} is starting.")
    def stop_engine(self):
        print(f"The engine of {self.year} {self.brand} {self.model} is stopping.")
    def display_info(self):
        print("Car Brand:",self.brand)
        print("Car Model:",self.model)
        print("Car Year:",self.year)

bmw = car("BMW", "X5", 2022)
bmw.display_info()
bmw.start_engine()
bmw.stop_engine()
```
