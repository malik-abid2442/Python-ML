# List Comprehension

```python
sqr = [i**2 for i in range(10)]
sqr
```

```python
odd_sqr = [i**2 for i in range(10) if i%2!=0]
odd_sqr
```

```python
nums = [3, 8, 11, 14, 17, 20, 25, 30]
even = [i for i in nums if i%2==0]
even
```

```python
words = ["apple", "cat", "banana", "dog", "elephant", "sun"]
length = [word for word in words if len(word)>5]
length
```

```python
words = ["python", "data", "science", "ml"]
upper_word = [word.upper() for word in words]
upper_word
```

```python
nums = [-5, 10, -2, 8, 0, -1, 15]
pos = [num for num in nums if num>= 0]
pos
```

```python
nums = [1, 2, 3, 4, 5, 6]
num = ["EVEN" if num%2==0 else "ODD" for num in nums]
num
```

```python
name = "MachineLearning"
l1 = [i for i in name]
l1
```

```python
text = "ArtificialIntelligence"
l1 = [i for i in text if i.upper() in "AEIOU"]
l1
```

```python
words = ["python", "data", "science", "machine", "learning"]
l1 = [len(char) for char in words]
l1
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
l1 = [i for i in matrix]
l1
```

```python
numbers = [num for num in range(100) if (num%3==0 and num%5==0)]
numbers
```

```python
l1 = [5,10,15,20,25]
numbers = [num for num in l1 if num>15]
numbers 
```
